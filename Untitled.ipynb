```python
import pandas as pd

# Sample DataFrame with a column to be normalized
data = {'column_to_normalize': [2, 4, 6, -1, 0, -5]}
df = pd.DataFrame(data)

# Min-Max scaling to normalize the column within the range [-1, 1]
min_val = df['column_to_normalize'].min()
max_val = df['column_to_normalize'].max()

df['normalized_column'] = ((df['column_to_normalize'] - min_val) / (max_val - min_val)) * 2 - 1

print(df)
```

```python
import pandas as pd

# Create a sample DataFrame
data = {'values': [-40, -0.5, -0.2, 0, 0.3, 0.7]}
df = pd.DataFrame(data)

# Define a custom function to transform the values
v=df['values']
def transform_range(x):
    return -1 + 2 * (x - v.min()) / (v.max() - v.min());

# Apply the custom function to the 'values' column
df['transformed_values'] = df['values'].apply(transform_range)

print(df)
```

```python
import pandas as pd

# Create a sample DataFrame
data = {'values': [-50, -10, 0, 5, 10, 0 ]}
df = pd.DataFrame(data)

# Define a custom function to transform the values
v_min = df['values'].min()
v_max = df['values'].max()

def transform_range(x):
    if x < 0:
        return (abs(x)) / (abs(v_min)) * -1
    elif x > 0:
        return x/v_max
    else:
        return 0

# Apply the custom function to the 'values' column
df['transformed_values'] = df['values'].apply(transform_range)

print(df)
```
