```python
import pandas as pd 
import numpy as np 
pd.set_option('display.expand_frame_repr', False)
pd.set_option('display.max_columns', None)
pd.set_option('display.expand_frame_repr', False)
pd.set_option('max_colwidth', -1)
```

```python
KNAB_donations_raw = pd.read_excel("20240125_KNAB_donations.xlsx")
```

```python
beneficial_owners_raw = pd.read_csv("beneficial_owners.csv", delimiter=';')
```

```python
beneficial_owners = beneficial_owners_raw.copy()
KNAB_donations = KNAB_donations.copy()
```

```python
beneficial_owners['PrimaryKey'] = beneficial_owners['forename'] + " " + beneficial_owners['surname'] +  beneficial_owners['latvian_identity_number_masked']
beneficial_owners['PrimaryKey'] = beneficial_owners['PrimaryKey'].str.replace(r'\W', '', regex=True)

beneficial_owners_df_count = beneficial_owners.groupby(['PrimaryKey']).agg('count').reset_index()[['PrimaryKey', 'legal_entity_registration_number']]
beneficial_owners_df_count.columns = ['PrimaryKey', 'NumberOfLegalEntities']
beneficial_owners_df_count
```

```python
KNAB_donations['PrimaryKey'] = KNAB_donations['donor'].str.replace(r'\W', '', regex=True)
```

```python
beneficial_owners
```

```python
businessOwnersWhoDonateToPoliticalParties = KNAB_donations.merge(beneficial_owners_df_count, on='PrimaryKey', how='inner')

businessOwnersWhoDonateToPoliticalParties_agg = businessOwnersWhoDonateToPoliticalParties.groupby(['PrimaryKey']).agg({'NumberOfLegalEntities':['max'],'amount_EUR_modif':['sum'],'party_regnum':['count', 'nunique'],})
```

```python
businessOwnersWhoDonateToPoliticalParties_agg
```

```python
businessOwnersWhoDonateToPoliticalParties_agg.columns = businessOwnersWhoDonateToPoliticalParties_agg.columns.get_level_values(0)
```

```python
businessOwnersWhoDonateToPoliticalParties_agg
```

```python
businessOwnersWhoDonateToPoliticalParties_agg.columns = ['NumberOfLegalEntities', 'amount_EUR_modif', 'Number_of_times_donated', 'Unique_parties_donated']
```

```python
cond1 = businessOwnersWhoDonateToPoliticalParties_agg['Unique_parties_donated'] >2
cond2 = businessOwnersWhoDonateToPoliticalParties_agg['amount_EUR_modif'] >10000
```
